## 2. 两数之和
给定 `nums = [2, 7, 11, 15]`、`target = 9`。先写暴力版，再写 dict 版。

````python
nums = [2, 7, 11, 15]
target = 9
# 暴力版本
````

````python
nums = [2, 7, 11, 15]
target = 9
# dict 版本
````

<details><summary>参考答案</summary>

```python
def two_sum(nums, target):
    seen = {}
    for i, x in enumerate(nums):
        need = target - x
        if need in seen:
            return [seen[need], i]
        seen[x] = i
```

## 3. 移动零（基础版）
给定 `nums = [0, 1, 0, 3, 12]`，用额外列表得到 `[1, 3, 12, 0, 0]`。

````python
nums = [0, 1, 0, 3, 12]
# 在这里写代码
````

<details><summary>参考答案</summary>

```python
result = []
for x in nums:
    if x != 0:
        result.append(x)
while len(result) < len(nums):
    result.append(0)
print(result)
```

## 4. 只出现一次的数字（计数版）
给定 `nums = [4, 1, 2, 1, 2]`，用 dict 计数，找只出现一次的数字。

````python
nums = [4, 1, 2, 1, 2]
# 在这里写代码
````

<details><summary>参考答案</summary>

```python
count = {}
for x in nums:
    count[x] = count.get(x, 0) + 1
for x, freq in count.items():
    if freq == 1:
        print(x)  # 4
        break
```
</details>
